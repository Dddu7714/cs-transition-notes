# Step 1 - Candidate Generation with RAPIDS
For candidate generation, we build three co-visitation matrices. One computes the popularity of cart/order given a user's previous click/cart/order. We apply type weighting to this matrix. One computes the popularity of cart/order given a user's previous cart/order. We call this "buy2buy" matrix. One computes the popularity of clicks given a user previously click/cart/order.  We apply time weighting to this matrix. We will use RAPIDS cuDF GPU to compute these matrices quickly!

```python
# 定义当前协同过滤结果文件的版本号，便于区分不同实验输出
VER = 6

# 导入 pandas 和 numpy，用于 CPU 端数据处理与数值计算
import pandas as pd, numpy as np
# 导入 notebook 版 tqdm，用于显示循环进度条
from tqdm.notebook import tqdm
# 导入文件系统、序列化、通配与垃圾回收等基础库
import os, sys, pickle, glob, gc
# 导入 Counter，用于计数统计
from collections import Counter
# 导入 cudf 以便在 GPU 上处理表格数据，并导入 itertools 处理迭代逻辑
import cudf, itertools
# 打印当前 RAPIDS 版本，方便确认运行环境
print('We will use RAPIDS version',cudf.__version__)
```

## Compute Three Co-visitation Matrices with RAPIDS
We will compute 3 co-visitation matrices using RAPIDS cuDF on GPU. This is 30x faster than using Pandas CPU like other public notebooks! For maximum speed, set the variable `DISK_PIECES` to the smallest number possible based on the GPU you are using without incurring memory errors. If you run this code offline with 32GB GPU ram, then you can use `DISK_PIECES = 1` and compute each co-visitation matrix in almost 1 minute! Kaggle's GPU only has 16GB ram, so we use `DISK_PIECES = 4` and it takes an amazing 3 minutes each! Below are some of the tricks to speed up computation
* Use RAPIDS cuDF GPU instead of Pandas CPU
* Read disk once and save in CPU RAM for later GPU multiple use
* Process largest amount of data possible on GPU at one time
* Merge data in two stages. Multiple small to single medium. Multiple medium to single large.
* Write result as parquet instead of dictionary

```python
# 设置运行模式，本地模式读取本地原始数据目录
MODE = "local" # "kaggle"

# 如果在 Kaggle 环境运行，则使用 Kaggle 对应的数据路径
if MODE == "kaggle":
    # 指向 Kaggle 中所有 parquet 分片文件
    readpath = '../raw_data/*_parquet/*'
# 如果在本地环境运行，则使用本地缓存的数据路径
elif MODE == "local":
    # 指向本地 deotte_radek 版本的 parquet 分片文件
    readpath = '../raw_data/deotte_radek/*_parquet/*'

# 使用 glob 收集所有待处理文件的完整路径
files = glob.glob(readpath)
```

```python
# 定义从 CPU 缓存中读取单个文件并转成 cudf.DataFrame 的函数
# 这里通过 GPU DataFrame 包装缓存数据，避免重复从磁盘读取
# 参数 f 表示单个 parquet 文件路径
# 返回值是该文件对应的 cudf.DataFrame
# 这样后续批处理时可以直接从内存拿数据
# 函数本身很轻量，仅负责格式转换
# 具体缓存填充逻辑由下方 read_file_to_cache 完成
# 这是整个协同访问矩阵构建流程的基础读取接口
# 通过函数封装也方便后续切换读取策略
# 注意这里依赖外部变量 data_cache
# 因此调用前必须先完成缓存初始化
# GPU 端读取相比反复读 parquet 更快
# 这里返回的是按文件粒度缓存的数据
# 有助于后续分块组合
# 适合大规模 session 事件数据预处理
# 开始定义函数
# 从缓存中读取并构建 cudf DataFrame
# 返回 GPU 版本的数据表
# 函数定义结束
# CACHE FUNCTIONS
# 定义 GPU 端读取缓存的函数
# 参数 f 是文件路径键
# 返回值是对应缓存内容构成的 cudf DataFrame
# 这样可以减少重复 IO
# 封装后便于后续代码复用
# 通过 data_cache 统一管理原始分片数据
# 让外层循环只关心计算逻辑
# 使用 cudf.DataFrame 便于后续 merge/groupby 在 GPU 上执行
# 这里不会修改缓存本身
# 每次调用都重新构造一个 cudf 视图
# 可视为 CPU 到 GPU 的转换桥梁
# 函数开始
# 返回指定文件的 GPU DataFrame
# 函数结束

def read_file(f):
    # 从预加载的缓存字典中取出指定文件的数据，并转换为 cudf.DataFrame
    return cudf.DataFrame( data_cache[f] )

# 定义读取 parquet 文件并写入 CPU 缓存的函数
# 该函数负责基础字段清洗与类型压缩
# 参数 f 为 parquet 文件路径
# 返回值是处理后的 pandas DataFrame
# 后续这些 DataFrame 会统一保存在 data_cache 中
# 先在 CPU 端处理可以降低 GPU 显存压力
# 同时减少重复的数据类型转换开销
# 时间戳和行为类型都会在这里标准化
# 这一步属于全流程的一次性预处理
# 函数开始

def read_file_to_cache(f):
    # 使用 pandas 读取单个 parquet 文件到内存
    df = pd.read_parquet(f)
    # 将毫秒级时间戳转换为秒级 int32，减小内存占用并方便后续时间窗计算
    df.ts = (df.ts/1000).astype('int32')
    # 将行为类型从字符串映射为整数编码，并压缩为 int8
    df['type'] = df['type'].map(type_labels).astype('int8')
    # 返回清洗后的单文件数据
    return df

# CACHE THE DATA ON CPU BEFORE PROCESSING ON GPU
# 初始化用于保存所有文件数据的 CPU 缓存字典
# 键是文件路径，值是预处理后的 pandas DataFrame
# 这样后续反复遍历数据时不用每次重新读磁盘
# 对大规模 parquet 分片非常重要
# 可以显著减少 IO 开销
# 同时让 GPU 端只负责计算密集逻辑
# 开始初始化缓存容器
# 创建空字典保存缓存数据
# 后续会逐文件填充
# 初始化完成

data_cache = {}
# 定义行为类型到整数标签的映射关系
# clicks、carts、orders 分别映射到 0、1、2
# 数值化后更适合后续权重映射和 GPU 计算
# 这里是整个项目通用的行为编码规则
# 供上面的 read_file_to_cache 使用
# 定义类型映射字典
# 点击映射为 0，加购映射为 1，下单映射为 2
# 方便后续统一处理

type_labels = {'clicks':0, 'carts':1, 'orders':2}
# 遍历所有 parquet 文件并提前读入 CPU 缓存
# 每个文件都会先做字段标准化和类型压缩
# 这样真正构建共现矩阵时就不需要重复预处理
# 循环开始
for f in files:
    # 将当前文件预处理后写入缓存字典
    data_cache[f] = read_file_to_cache(f)

# CHUNK PARAMETERS
# 设置每次内层批处理要一起读取的文件数
# 文件过多会占用过多显存，过少又会增加调度开销
# 这里取 5 作为经验值
READ_CT = 5
# 将全部文件划分为 6 个外层块，并计算每个外层块的大小
CHUNK = int( np.ceil( len(files)/6 ))
# 打印当前文件总数、每组读取数和外层块大小，便于观察任务规模
print(f'We will process {len(files)} files, in groups of {READ_CT} and chunks of {CHUNK}.')
```

## 1) "Carts Orders" Co-visitation Matrix - Type Weighted

```python
# 定义不同行为类型在共现统计中的权重
# 点击权重较低，加购和下单权重更高，体现购买意图更强
# 这里构建的是 carts/orders 相关的协同访问矩阵
# 因此使用加权策略增强高价值行为信号
# 定义类型权重映射字典
# clicks 权重为 1
# carts 权重为 5
# orders 权重为 4
# 供后续 type_y.map 使用
# 权重配置完成

type_weight = {0:1, 1:5, 2:4}

# USE SMALLEST DISK_PIECES POSSIBLE WITHOUT MEMORY ERROR
# 将 aid_x 空间切成多个磁盘分块，以控制单次计算的显存占用
DISK_PIECES = 16
# 根据物品总量估算每个分块覆盖的 aid 区间大小
SIZE = 1.86e6/DISK_PIECES

# COMPUTE IN PARTS FOR MEMORY MANGEMENT
# 遍历每个磁盘分块，分别构建局部共现矩阵并保存
for PART in range(DISK_PIECES):
    # 打印空行，便于观察日志输出
    print()
    # 打印当前正在处理的磁盘分块编号
    print('### DISK PART',PART+1)
    
    # MERGE IS FASTEST PROCESSING CHUNKS WITHIN CHUNKS
    # => OUTER CHUNKS
    # 外层将全部文件按 6 大块分段处理，降低单次合并规模
    for j in range(6):
        # 计算当前外层块的起始文件索引
        a = j*CHUNK
        # 计算当前外层块的结束文件索引上界
        b = min( (j+1)*CHUNK, len(files) )
        # 打印当前外层块覆盖的文件范围和内层组大小
        print(f'Processing files {a} thru {b-1} in groups of {READ_CT}...')
        
        # => INNER CHUNKS
        # 在当前外层块内部，再按 READ_CT 个文件一组读取和统计
        for k in range(a,b,READ_CT):
            # READ FILE
            # 先读取当前组的第一个文件
            df = [read_file(files[k])]
            # 继续尝试将同组后续文件一起加入列表
            for i in range(1,READ_CT): 
                # 如果后续文件索引没有越界，则继续追加读取
                if k+i<b: df.append( read_file(files[k+i]) )
            # 将当前小组内多个文件按行拼接成一个大表
            df = cudf.concat(df,ignore_index=True,axis=0)
            # 按 session 升序、时间戳降序排序，保证最近行为排在前面
            df = df.sort_values(['session','ts'],ascending=[True,False])
            
            # USE TAIL OF SESSION
            # 重置索引，便于后续按位置生成会话内顺序号
            df = df.reset_index(drop=True)
            # 对每个 session 计算排序后的行为序号
            df['n'] = df.groupby('session').cumcount()
            # 仅保留每个 session 最近 30 条行为，聚焦短期兴趣并控制计算量
            df = df.loc[df.n<30].drop('n',axis=1)
            
            # CREATE PAIRS
            # 按 session 自连接，生成同一会话内物品两两配对
            df = df.merge(df,on='session')
            # 保留 24 小时内发生的共现对，并排除同一物品自身配对
            df = df.loc[ ((df.ts_x - df.ts_y).abs()< 24 * 60 * 60) & (df.aid_x != df.aid_y) ]
            
            # MEMORY MANAGEMENT COMPUTE IN PARTS
            # 只保留当前 PART 对应 aid_x 区间的数据，分块降低显存压力
            df = df.loc[(df.aid_x >= PART*SIZE)&(df.aid_x < (PART+1)*SIZE)]
            
            # ASSIGN WEIGHTS
            # 在 session + aid_x + aid_y + type_y 粒度去重，避免同类重复事件重复加权
            df = df[['session', 'aid_x', 'aid_y','type_y']].drop_duplicates(['session', 'aid_x', 'aid_y', 'type_y'])
            # 根据目标行为类型 type_y 映射对应权重
            df['wgt'] = df.type_y.map(type_weight)
            # 只保留共现对及其权重字段
            df = df[['aid_x','aid_y','wgt']]
            # 将权重压缩为 float32，减少显存占用
            df.wgt = df.wgt.astype('float32')
            # 对每个 aid_x, aid_y 共现对累计权重
            df = df.groupby(['aid_x','aid_y']).wgt.sum()
            
            # COMBINE INNER CHUNKS
            # 如果是当前外层块中的第一组，则直接初始化局部结果
            if k==a: tmp2 = df
            # 否则将当前组结果累加到外层块结果中
            else: tmp2 = tmp2.add(df, fill_value=0)
            # 打印当前处理到的起始文件索引，便于跟踪进度
            print(k,', ',end='')
        
        # 当前外层块处理完成后换行输出
        print()
        
        # COMBINE OUTER CHUNKS
        # 如果是第一个外层块，则直接初始化全局结果
        if a==0: tmp = tmp2
        # 否则把当前外层块的结果累加到总结果中
        else: tmp = tmp.add(tmp2, fill_value=0)
        # 删除中间变量释放显存/内存
        del tmp2, df
        # 主动触发垃圾回收
        gc.collect()

    # CONVERT MATRIX TO DICTIONARY
    # 将分组结果从索引形式还原为普通表结构
    tmp = tmp.reset_index()
    # 对每个 aid_x 按权重从高到低排序
    tmp = tmp.sort_values(['aid_x','wgt'],ascending=[True,False])
    
    # SAVE TOP 40
    # 重置索引，方便对每个 aid_x 重新计数排名
    tmp = tmp.reset_index(drop=True)
    # 计算每个 aid_x 下 aid_y 的排序位置
    tmp['n'] = tmp.groupby('aid_x').aid_y.cumcount()
    # 每个 aid_x 只保留前 50 个最强共现候选
    tmp = tmp.loc[tmp.n<50].drop('n',axis=1)
    
    # SAVE PART TO DISK (convert to pandas first uses less memory)
    # 将当前分块结果转回 pandas 后写入 parquet 文件
    tmp.to_pandas().to_parquet(f'{MODE}_top_50_carts_orders_v{VER}_{PART}.pqt')
```

## 2) "Buy2Buy" Co-visitation Matrix

```python
# USE SMALLEST DISK_PIECES POSSIBLE WITHOUT MEMORY ERROR
# buy2buy 矩阵只切 1 份，因为过滤到 carts/orders 后数据量相对更可控
DISK_PIECES = 1
# 根据分块数计算当前 aid 区间大小
SIZE = 1.86e6/DISK_PIECES

# COMPUTE IN PARTS FOR MEMORY MANGEMENT
# 遍历每个磁盘分块，构建 buy2buy 共现矩阵
for PART in range(DISK_PIECES):
    # 打印空行便于分隔日志
    print()
    # 打印当前分块编号
    print('### DISK PART',PART+1)
    
    # MERGE IS FASTEST PROCESSING CHUNKS WITHIN CHUNKS
    # => OUTER CHUNKS
    # 外层仍按 6 个大块处理文件
    for j in range(6):
        # 计算当前外层块起始文件索引
        a = j*CHUNK
        # 计算当前外层块结束文件索引上界
        b = min( (j+1)*CHUNK, len(files) )
        # 输出当前处理范围
        print(f'Processing files {a} thru {b-1} in groups of {READ_CT}...')
        
        # => INNER CHUNKS
        # 在外层块内每次读取 READ_CT 个文件
        for k in range(a,b,READ_CT):
            
            # READ FILE
            # 先读取当前小组的第一个文件
            df = [read_file(files[k])]
            # 继续将本组其他文件加入
            for i in range(1,READ_CT): 
                # 只要索引未越界就追加读取
                if k+i<b: df.append( read_file(files[k+i]) )
            # 将本组多个文件按行拼接
            df = cudf.concat(df,ignore_index=True,axis=0)
            # buy2buy 只关注加购和下单行为，因此过滤掉点击
            df = df.loc[df['type'].isin([1,2])] # ONLY WANT CARTS AND ORDERS
            # 仍按 session 升序、时间戳降序排列
            df = df.sort_values(['session','ts'],ascending=[True,False])
            
            # USE TAIL OF SESSION
            # 重置索引，准备生成会话内顺序号
            df = df.reset_index(drop=True)
            # 对每个 session 计算行为次序
            df['n'] = df.groupby('session').cumcount()
            # 仅保留最近 30 条行为
            df = df.loc[df.n<30].drop('n',axis=1)
            
            # CREATE PAIRS
            # 对同一 session 内的物品两两配对
            df = df.merge(df,on='session')
            # 保留 14 天内的共现对，并去掉自身配对
            df = df.loc[ ((df.ts_x - df.ts_y).abs()< 14 * 24 * 60 * 60) & (df.aid_x != df.aid_y) ] # 14 DAYS
            
            # MEMORY MANAGEMENT COMPUTE IN PARTS
            # 只保留当前 aid_x 分块范围内的数据
            df = df.loc[(df.aid_x >= PART*SIZE)&(df.aid_x < (PART+1)*SIZE)]
            
            # ASSIGN WEIGHTS
            # 在 session + aid_x + aid_y + type_y 粒度去重
            df = df[['session', 'aid_x', 'aid_y','type_y']].drop_duplicates(['session', 'aid_x', 'aid_y', 'type_y'])
            # buy2buy 中所有保留下来的行为统一赋权为 1
            df['wgt'] = 1
            # 只保留共现对和权重
            df = df[['aid_x','aid_y','wgt']]
            # 将权重转为 float32
            df.wgt = df.wgt.astype('float32')
            # 按共现对累加权重
            df = df.groupby(['aid_x','aid_y']).wgt.sum()
            
            # COMBINE INNER CHUNKS
            # 第一组直接初始化外层块结果
            if k==a: tmp2 = df
            # 其余小组累加到外层块结果中
            else: tmp2 = tmp2.add(df, fill_value=0)
            # 输出当前组进度
            print(k,', ',end='')

        # 当前外层块结束后换行
        print()
        
        # COMBINE OUTER CHUNKS
        # 第一外层块直接作为总结果初始化
        if a==0: tmp = tmp2
        # 其他外层块继续累加
        else: tmp = tmp.add(tmp2, fill_value=0)
        # 删除中间变量并回收内存
        del tmp2, df
        gc.collect()

    # CONVERT MATRIX TO DICTIONARY
    # 将分组结果还原为普通表结构
    tmp = tmp.reset_index()
    # 按 aid_x 分组后根据权重降序排列 aid_y
    tmp = tmp.sort_values(['aid_x','wgt'],ascending=[True,False])
    
    # SAVE TOP 40
    # 重置索引以便重新编号排序位次
    tmp = tmp.reset_index(drop=True)
    # 计算每个 aid_x 下候选 aid_y 的排名
    tmp['n'] = tmp.groupby('aid_x').aid_y.cumcount()
    # 每个 aid_x 只保留前 50 个候选
    tmp = tmp.loc[tmp.n<50].drop('n',axis=1)
    
    # SAVE PART TO DISK (convert to pandas first uses less memory)
    # 将 buy2buy 结果写出为 parquet 文件
    tmp.to_pandas().to_parquet(f'{MODE}_top_50_buy2buy_v{VER}_{PART}.pqt')
```

## 3) "Clicks" Co-visitation Matrix - Time Weighted

```python
# USE SMALLEST DISK_PIECES POSSIBLE WITHOUT MEMORY ERROR
# clicks 协同矩阵切成 4 份，以平衡显存占用与运行速度
DISK_PIECES = 4
# 根据分块数估算每份对应的 aid 范围大小
SIZE = 1.86e6/DISK_PIECES

# COMPUTE IN PARTS FOR MEMORY MANGEMENT
# 遍历每个磁盘分块，构建 clicks 协同矩阵
for PART in range(DISK_PIECES):
    # 打印空行方便阅读日志
    print()
    # 打印当前分块编号
    print('### DISK PART',PART+1)
    
    # MERGE IS FASTEST PROCESSING CHUNKS WITHIN CHUNKS
    # => OUTER CHUNKS
    # 外层按 6 个大块遍历全部文件
    for j in range(6):
        # 计算当前外层块起始文件索引
        a = j*CHUNK
        # 计算当前外层块结束文件索引上界
        b = min( (j+1)*CHUNK, len(files) )
        # 打印当前处理范围
        print(f'Processing files {a} thru {b-1} in groups of {READ_CT}...')
        
        # => INNER CHUNKS
        # 在外层块内部按 READ_CT 个文件进行批处理
        for k in range(a,b,READ_CT):
            # READ FILE
            # 读取当前小组第一个文件
            df = [read_file(files[k])]
            # 追加读取同组后续文件
            for i in range(1,READ_CT): 
                # 如果索引未越界则继续追加
                if k+i<b: df.append( read_file(files[k+i]) )
            # 将当前小组文件拼接成一个大表
            df = cudf.concat(df,ignore_index=True,axis=0)
            # 按 session 升序、时间戳降序排列
            df = df.sort_values(['session','ts'],ascending=[True,False])
            
            # USE TAIL OF SESSION
            # 重置索引，准备生成会话内顺序号
            df = df.reset_index(drop=True)
            # 对每个 session 计算行为顺序
            df['n'] = df.groupby('session').cumcount()
            # 只保留每个 session 最近 30 条记录
            df = df.loc[df.n<30].drop('n',axis=1)
            
            # CREATE PAIRS
            # 在同一 session 内做物品两两配对
            df = df.merge(df,on='session')
            # 保留 24 小时内的共现对，并排除同物品配对
            df = df.loc[ ((df.ts_x - df.ts_y).abs()< 24 * 60 * 60) & (df.aid_x != df.aid_y) ]
            
            # MEMORY MANAGEMENT COMPUTE IN PARTS
            # 只保留当前 aid_x 分块范围内的数据
            df = df.loc[(df.aid_x >= PART*SIZE)&(df.aid_x < (PART+1)*SIZE)]
            
            # ASSIGN WEIGHTS
            # 在 session + aid_x + aid_y 粒度去重，避免重复计数
            df = df[['session', 'aid_x', 'aid_y','ts_x']].drop_duplicates(['session', 'aid_x', 'aid_y'])
            # 根据事件时间给近期行为更高权重，越新的行为贡献越大
            df['wgt'] = 1 + 3*(df.ts_x - 1659304800)/(1662328791-1659304800)
            # 1659304800 : minimum timestamp
            # 1662328791 : maximum timestamp
            # 只保留共现对及其权重
            df = df[['aid_x','aid_y','wgt']]
            # 将权重转为 float32 以节省显存
            df.wgt = df.wgt.astype('float32')
            # 对每个共现对累计时间衰减后的权重
            df = df.groupby(['aid_x','aid_y']).wgt.sum()
            
            # COMBINE INNER CHUNKS
            # 第一组直接初始化当前外层块结果
            if k==a: tmp2 = df
            # 其余小组将结果累加进去
            else: tmp2 = tmp2.add(df, fill_value=0)
            # 打印当前处理进度
            print(k,', ',end='')
        # 当前外层块结束后换行
        print()
        
        # COMBINE OUTER CHUNKS
        # 第一外层块直接初始化总结果
        if a==0: tmp = tmp2
        # 其余外层块继续累加
        else: tmp = tmp.add(tmp2, fill_value=0)
        # 删除中间变量并回收内存
        del tmp2, df
        gc.collect()

    # CONVERT MATRIX TO DICTIONARY
    # 将聚合结果恢复为普通表结构
    tmp = tmp.reset_index()
    # 对每个 aid_x 下的 aid_y 候选按权重降序排列
    tmp = tmp.sort_values(['aid_x','wgt'],ascending=[True,False])
    
    # SAVE TOP 40
    # 重置索引方便重新计算组内排名
    tmp = tmp.reset_index(drop=True)
    # 统计每个 aid_x 下 aid_y 的排序位置
    tmp['n'] = tmp.groupby('aid_x').aid_y.cumcount()
    # 每个 aid_x 仅保留前 50 个点击共现候选
    tmp = tmp.loc[tmp.n<50].drop('n',axis=1)
    
    # SAVE PART TO DISK (convert to pandas first uses less memory)
    # 将 clicks 共现结果写出到 parquet 文件
    tmp.to_pandas().to_parquet(f'{MODE}_top_50_clicks_v{VER}_{PART}.pqt')
```
